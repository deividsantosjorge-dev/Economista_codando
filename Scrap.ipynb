```python
import requests
from bs4 import BeautifulSoup

URL = "https://books.toscrape.com/"
res = requests.get(URL)
print("status_code:", res.status_code)

soup = BeautifulSoup(res.text)

```

```python
livros = soup.find_all("article", class_="product_pod")
##print("Livros nesta página:", len(livros))

print("livros")
```

```python
dados = []
for livro in livros:
    titulo = livro.find("h3").find("a")["title"]          # atributo title
    preco  = livro.find("p", class_="price_color").get_text(strip=True)
    # A classe star-rating tem 2 valores, ex.: ["star-rating", "Three"]
    nota = livro.find("p", class_="star-rating")["class"][1]
    estoque=livro.find("p", class_="availability").get_text(strip=True)


    #nota   = livro.find("p", class_="star-rating")["class"][1]
    #estoque = livro.find("p", class_="instock").get_text(strip=True)

    dados.append({"titulo": titulo, "preco": preco, "nota": nota, "estoque": estoque})
    print(f"{nota:5} | {preco} | {titulo:8} | {estoque:20}")
```

```python

# Em seguida, importo a biblioteca BeautifulSoup,
# que será usada para interpretar e navegar pelo HTML recebido.
from bs4 import BeautifulSoup

# Transformo o HTML armazenado em reque.text em um objeto BeautifulSoup.
# O "html.parser" é o analisador que organiza esse HTML para que possamos localizar tags, atributos e textos.
#Basicamente, no BeautifulSoup a regra é: 1º argumento → conteúdo que será analisado e 2º argumento → parser que será utilizado
soup = BeautifulSoup(reque.text, "html.parser")

# Procuro no HTML todos os elementos que possuem data-hook="reviewRichContentContainer".
# O select() retorna uma lista com todos os elementos encontrados.
comentarios = soup.select('[data-hook="reviewRichContentContainer"]')

# Verifico quantos comentários foram encontrados.
print(len(comentarios))
```

```python
# Percorro cada elemento encontrado na lista "comentarios".
# enumerate() também cria uma numeração para cada comentário,
# começando em 1 por causa de start=1.
for i, comentario in enumerate(comentarios, start=1):

    # Extraio apenas o texto existente dentro do elemento HTML.
    # strip=True remove espaços e quebras de linha desnecessários.
    print(i, comentario.get_text(strip=True))

    # Imprime uma linha para separar visualmente os comentários.
    print("------------")
```
